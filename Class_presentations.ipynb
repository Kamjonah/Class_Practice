```python
#lambda structure
#lambda arguments: expression
```

```python
add = lambda x: x+3
print(add(5))
```

```python
double = lambda x: x*x
print(double(2))
```

```python
#combining lambda function with if function
num = lambda x: "positive" if x>0 else "negative"
print(num(-3))
```

```python
temperature = lambda degree: "hot" if degree>30 else "cold" if degree>=35 else "very hot"
print(temperature(39))
```

```python
# Recursion
# Countdown
# factorial
```

```python
def factorial(n)
  if n == 1
    return 1
  return n * factorial
```

```python
def tail_fact(n, acc=1):
  if n == 0:
    return acc
  else:
    return tail_fact(n-1, n*acc)
print(tail_fact(5))
```
